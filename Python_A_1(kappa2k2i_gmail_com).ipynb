***Answer to the question no 01***


```python
lst = list(input().split())
ext_num = []
for i in range(len(lst)):
  if int(lst[i]) % 4 == 0 and int(lst[i]) % 8 != 0:
    ext_num.append(int(lst[i]))

print(ext_num)
```

***Answer to the question no 02***

```python
def compute_ratio(x, y):
  try:
    result = (x + y) / (x * y)
    return result
  except ZeroDivisionError:
    return "Error: Division not possible"

compute_ratio(5, 0)
```

***Answer to the question no 03***

```python
with open("./sample_data/data.txt", "r") as file:
  text = file.read()
  words = text.split()

  longest = ""
  for word in words:
    if len(word) > len(longest):
      longest = word

  print(longest)
```

***Answer to the question no 04***

```python
sentence = input()
lst = sentence.split()

new_lst = []
for word in lst:
  if word[0].lower() in "aeiou":
    new_lst.append(word)
print(new_lst)
```

***Answer to the question no 05***

```python
word = input().lower()
r_word = word[::-1]

if word == r_word:
  print("Palindrome")
else:
  print("Not Palindrome")
```
